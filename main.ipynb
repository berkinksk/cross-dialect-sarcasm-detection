# Run all notebooks

The cells below run each notebook in its own kernel with `jupyter nbconvert --execute`, which avoids package version conflicts between sections. Outputs are saved back into each notebook.

If the cell above installed `sentencepiece` or `protobuf`, restart the kernel once before continuing.

```python
import subprocess
from pathlib import Path

notebooks = [
    'Q1_data_analysis/Q1_1_Data_Distribution.ipynb',
    'Q1_data_analysis/Q1_2_Vocabulary_Analysis.ipynb',
    'Q2_1_baselines/Q2_1_Baseline_SVM_and_RoBERTa_and_DeBERTa.ipynb',
    'Q2_2_cross_variety/Q2_2_Cross_Variety_Evaluation.ipynb',
    'Q2_2_cross_variety/Q2_2_DeBERTa_Combined.ipynb',
    'Q2_3_lora_llms/main_q2_3.ipynb',
    'Q4_error_analysis/Q4_Sarcasm_Error_Analysis_DeBERTa.ipynb',
    'Q4_error_analysis/Q4_Sarcasm_Error_Analysis_SmolLM2-1.7B+LoRA.ipynb',
    'Q5_2_efficiency/Q5_2_Efficiency.ipynb',
]

def run_nb(path):
    print(f'\n=== {path} ===')
    if not Path(path).exists():
        print('skip (file not found)')
        return
    cmd = ['jupyter', 'nbconvert', '--to', 'notebook', '--execute', '--inplace', path]
    result = subprocess.run(cmd, capture_output=True, text=True)
    if result.returncode == 0:
        print('done')
    else:
        print('error')
        print(result.stderr[-1500:])
```

```python
for nb in notebooks:
    run_nb(nb)

print('\nall notebooks attempted. open each one to see its outputs.')
```
